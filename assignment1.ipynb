```python
#17. Write a program to check grade based on marks (A, B, C, Fail).

marks1 = int(input("Enter 1 sub marks: "))
marks2 = int(input("Enter 2 sub marks: "))
marks3 = int(input("Enter 3 sub marks: "))

per = (marks1 + marks2 + marks3) / 3

print("Percentage:", per)

if per >= 80:
  print("Grade: A")
elif per >= 60:
  print("Grade: B")
elif per >= 40:
  print("Grade: C")
else:
  print("Grade: Oops!! Fail Try Again")
    
```

```python
#19. Write a program to print the day name based on day number (1–7).

day = int(input("Enter day number (1-7): "))

if day == 1:
  print("Monday")
elif day == 2:
  print("Tuesday")
elif day == 3:
  print("Wednesday")
elif day == 4:
  print("Thursday")
elif day == 5:
  print("Friday")
elif day == 6:
  print("Saturday")
elif day == 7:
  print("Sunday")
else:
  print("Invalid day number! Please enter between 1 and 7.")
```

```python
day = int(input("Enter day number (1-7): "))

match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
  case _:
    print("Invalid day number! Please enter between 1 and 7.")
```
